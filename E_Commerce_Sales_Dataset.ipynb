```python
ECom_clean = ECom_sales[~ ECom_sales['Courier Status'].isin(['Cancelled' , 'Unshipped'])]

print(f'Missing Amount After: {ECom_clean['Amount'].isnull().sum()}')
print(f'Missing rate: {(ECom_clean['Amount'].isnull().sum() / len(ECom_clean) * 100):.2f}%')
```

```python
print('=== DATA CLEANING DECISION LOG ===\n')


print('Column: promotion-ids')
print(f'Non-null: {ECom_clean['promotion-ids'].isnull().sum()}/{len(ECom_clean)} (68.5%)')
print(' Decision: Drop - Not essential for core sales analysis')
print(' Reason: For this project , analyzing overall sales trend > promo effectiveness\n')

print('Column: fulfilled-by')
print(f'Non-null: {ECom_clean['fulfilled-by'].isnull().sum()}/{len(ECom_clean)} (33.3%)')
print(' Decision: Drop - Too sparse (66.7% missing) & not core to analysis')
print(' Reason: Focus on revenue & customer behavior , not fulfillment channels\n')

ECom_final = ECom_clean.drop(columns=['promotion-ids' , 'fulfilled-by'])
print(f'Final dataset: {ECom_final.shape}')

```

```python
# Standardize all column names
ECom_final.columns = ECom_final.columns.str.strip()  # Remove spaces
ECom_final.columns = ECom_final.columns.str.lower()  # Make lowercase
ECom_final.columns = ECom_final.columns.str.replace(' ', '_')  # Replace spaces with underscores

# Now all your columns will be clean and consistent
print(ECom_final.columns)
```

```python
ECom_final.drop('unnamed:_22' , axis = 1 , inplace = True)
```

```python
print("Missing Values in Courier_Status:")
print(ECom_final['courier_status'].value_counts(dropna=False))
```

```python
ECom_final['courier_status'] = ECom_final['courier_status'].fillna(ECom_final.apply(lambda row: 'Cancelled' if row['status'] == 'Cancelled' else ECom_final['courier_status'].mode()[0], axis = 1))
print(f"Missing Value after fill: {ECom_final['courier_status'].isnull().sum()}")
```

```python
ECom_final['currency'].fillna('INR' , inplace = True)
```

```python
# Check how many missing amounts
print("Missing values in Amount column:")
print(f"Total missing: {ECom_final['amount'].isnull().sum()}")
print(f"Percentage: {(ECom_final['amount'].isnull().sum() / len(ECom_final) * 100):.2f}%")

# Check if they correlate with any status
missing_amount_data = ECom_final[ECom_final['amount'].isnull()]
print("\nMissing amounts by Status:")
print(missing_amount_data['status'].value_counts())
```

```python
print("=== AMOUNT CLEANUP ===")
before = len(ECom_final)
ECom_final = ECom_final.dropna(subset=['amount'])
after = len(ECom_final)

print(f"✅ Rows removed: {before - after}")
print(f"✅ Final dataset: {after} rows")
print(f"✅ Missing amounts: {ECom_final['amount'].isnull().sum()}")

# Final quality check
print(f"\n=== FINAL DATA QUALITY ===")
print(f"Missing values:\n{ECom_final.isnull().sum()}")
print(f"\nData Quality Score: {(1 - ECom_final.isnull().sum().sum() / (len(ECom_final) * len(ECom_final.columns))) * 100:.2f}%")
```

```python
## Revenue Metrics
Total_Revenue = ECom_final['amount'].sum()
print('Total_Revenue:' , Total_Revenue)
Avg_Order_Value = ECom_final['amount'].mean()
print('Avg_Order_Value:' , Avg_Order_Value)
```

```python
## Top Categories
Top_Categories = ECom_final.groupby('category')['amount'].sum().sort_values(ascending =False).head(10)
print('Top 10 Categories:' , Top_Categories)
```

```python
## B2B vs B2C
B2B_B2C = ECom_final.groupby('b2b')['amount'].agg(['sum', 'mean', 'count'])
print('B2B_Analysis:', B2B_B2C)
```

```python
## Top Products

Top_Products = ECom_final.groupby('sku')['amount'].sum().sort_values(ascending = False).head(10)
print('Top 10 Products:' , Top_Products)
```

```python
## Order Status Impact
Status_Analysis = ECom_final.groupby('status')['amount'].agg(['sum' , 'mean' , 'count'])
print('Status Analysis:', Status_Analysis )
```

```python
# Export cleaned data for Power BI
ECom_final.to_csv('ECom_final_cleaned.csv', index=False)
print("Data exported successfully!")
print(f"File size: {ECom_final.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
```

```python
print("=" * 60)
print("KEY INSIGHTS FROM ANALYSIS")
print("=" * 60)

print("\n1. TOP REVENUE DRIVER:")
print(f"   Category: {Top_Categories.index[0]}")
print(f"   Revenue: ${Top_Categories.values[0]:,.0f}")

print("\n2. B2B vs B2C:")
b2b_sum = B2B_B2C.loc[True, 'sum']
b2c_sum = B2B_B2C.loc[False, 'sum']
print(f"   B2C: ${b2c_sum:,.0f} ({b2c_sum/(b2b_sum+b2c_sum)*100:.1f}%)")
print(f"   B2B: ${b2b_sum:,.0f} ({b2b_sum/(b2b_sum+b2c_sum)*100:.1f}%)")

print("\n3. TOP PRODUCT:")
print(f"   SKU: {Top_Products.index[0]}")
print(f"   Revenue: ${Top_Products.values[0]:,.0f}")

print("\n4. ORDER STATUS:")
print(f"   Shipped %: {(Status_Analysis.loc['Shipped', 'sum'] / Status_Analysis['sum'].sum() * 100):.1f}%")
```
